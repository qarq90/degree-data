1.   Classsical set - a classical set is defined as in which the values are either fully in or fully out , a yes or no .

2.   Fuzzy set - a fuzzy set is defined as a set where the members of the set have a degree of membership.


Operations on Fuzzy set -

1.  Union = It is the maximum of the degree of membership of the two sets
2.  Intersection = It is the minimum of the degree of membership of the two sets .
1.  Complement = It means negation of the degree of membership i.e 1-degree of membership.
2.  Substraction = substraction of two sets a and be is indicated by a-b = a intersection ~b  




Importing the numpy library to perform operaions on the fuzzy set

```python
import numpy as np

A = np.array([0.4, 0.5, 0.7, 0.2, 1.0])
B = np.array([0.7, 0.3, 0.8, 0.6, 0.0])

elements = np.array([1, 2, 3, 4, 5])

print("\nMEMBERSHIP VALUES:")
print(f"Set A: {A[0]}/1, {A[1]}/2, {A[2]}/3, {A[3]}/4, {A[4]}/5")
print(f"Set B: {B[0]}/1, {B[1]}/2, {B[2]}/3, {B[3]}/4, {B[4]}/5")

union = np.maximum(A, B)
print("\nUNION (A U B):")
print(f"Formula: max(muA(x), muB(x))")
print(f"Result: {union[0]}/1, {union[1]}/2, {union[2]}/3, {union[3]}/4, {union[4]}/5")
print(f"Calculation: max(0.4,0.7)=0.7/1, max(0.5,0.3)=0.5/2, max(0.7,0.8)=0.8/3, max(0.2,0.6)=0.6/4, max(1.0,0.0)=1.0/5")

intersection = np.minimum(A, B)
print("\nINTERSECTION (A n B):")
print(f"Formula: min(muA(x), muB(x))")
print(f"Result: {intersection[0]}/1, {intersection[1]}/2, {intersection[2]}/3, {intersection[3]}/4, {intersection[4]}/5")
print(f"Calculation: min(0.4,0.7)=0.4/1, min(0.5,0.3)=0.3/2, min(0.7,0.8)=0.7/3, min(0.2,0.6)=0.2/4, min(1.0,0.0)=0.0/5")

complement_A = 1 - A
complement_B = 1 - B
print("\nCOMPLEMENT:")
print(f"Complement of A (A'): {complement_A[0]}/1, {complement_A[1]}/2, {complement_A[2]}/3, {complement_A[3]}/4, {complement_A[4]}/5")
print(f"Calculation: 1-0.4=0.6/1, 1-0.5=0.5/2, 1-0.7=0.3/3, 1-0.2=0.8/4, 1-1.0=0.0/5")
print(f"Complement of B (B'): {complement_B[0]}/1, {complement_B[1]}/2, {complement_B[2]}/3, {complement_B[3]}/4, {complement_B[4]}/5")
print(f"Calculation: 1-0.7=0.3/1, 1-0.3=0.7/2, 1-0.8=0.2/3, 1-0.6=0.4/4, 1-0.0=1.0/5")

subtraction = np.minimum(A, 1 - B)
print("\nSUBTRACTION (A - B):")
print(f"Formula: A n B' = min(muA(x), 1-muB(x))")
print(f"Result: {subtraction[0]}/1, {subtraction[1]}/2, {subtraction[2]}/3, {subtraction[3]}/4, {subtraction[4]}/5")
print(f"Calculation: min(0.4,0.3)=0.3/1, min(0.5,0.7)=0.5/2, min(0.7,0.2)=0.2/3, min(0.2,0.4)=0.2/4, min(1.0,1.0)=1.0/5")

algebraic_product = A * B
print("\nALGEBRAIC PRODUCT (A * B):")
print(f"Formula: muA(x) * muB(x)")
print(f"Result: {algebraic_product[0]:.2f}/1, {algebraic_product[1]:.2f}/2, {algebraic_product[2]:.2f}/3, {algebraic_product[3]:.2f}/4, {algebraic_product[4]:.2f}/5")
print(f"Calculation: 0.4*0.7=0.28/1, 0.5*0.3=0.15/2, 0.7*0.8=0.56/3, 0.2*0.6=0.12/4, 1.0*0.0=0.00/5")

algebraic_sum = A + B - (A * B)
print("\nALGEBRAIC SUM (A + B - A*B):")
print(f"Formula: muA(x) + muB(x) - muA(x)*muB(x)")
print(f"Result: {algebraic_sum[0]:.2f}/1, {algebraic_sum[1]:.2f}/2, {algebraic_sum[2]:.2f}/3, {algebraic_sum[3]:.2f}/4, {algebraic_sum[4]:.2f}/5")
print(f"Calculation: 0.4+0.7-0.28=0.82/1, 0.5+0.3-0.15=0.65/2, 0.7+0.8-0.56=0.94/3, 0.2+0.6-0.12=0.68/4, 1.0+0.0-0.0=1.00/5")

bounded_sum = np.minimum(1, A + B)
print("\nBOUNDED SUM (min(1, A+B)):")
print(f"Formula: min(1, muA(x) + muB(x))")
print(f"Result: {bounded_sum[0]:.1f}/1, {bounded_sum[1]:.1f}/2, {bounded_sum[2]:.1f}/3, {bounded_sum[3]:.1f}/4, {bounded_sum[4]:.1f}/5")
print(f"Calculation: min(1,1.1)=1.0/1, min(1,0.8)=0.8/2, min(1,1.5)=1.0/3, min(1,0.8)=0.8/4, min(1,1.0)=1.0/5")

bounded_product = np.maximum(0, A + B - 1)
print("\nBOUNDED PRODUCT (max(0, A+B-1)):")
print(f"Formula: max(0, muA(x) + muB(x) - 1)")
print(f"Result: {bounded_product[0]:.1f}/1, {bounded_product[1]:.1f}/2, {bounded_product[2]:.1f}/3, {bounded_product[3]:.1f}/4, {bounded_product[4]:.1f}/5")
print(f"Calculation: max(0,0.1)=0.1/1, max(0,-0.2)=0.0/2, max(0,0.5)=0.5/3, max(0,-0.2)=0.0/4, max(0,0.0)=0.0/5")
```

- min(0.5, 0.3) = 0.3/2
- min(0.7, 0.8) = 0.7/3
- min(0.2, 0.6) = 0.2/4
- min(1.0, 0.0) = 0.0/5


## 3. COMPLEMENT

**Formula:** 1 - μ(x)

### Complement of A (Ā)

**Result:** 0.6/1, 0.5/2, 0.3/3, 0.8/4, 0.0/5

**Calculation:**
- 1 - 0.4 = 0.6/1
- 1 - 0.5 = 0.5/2
- 1 - 0.7 = 0.3/3
- 1 - 0.2 = 0.8/4

### Complement of B (B̄)

**Result:** 0.3/1, 0.7/2, 0.2/3, 0.4/4, 1.0/5

**Calculation:**
- 1 - 0.7 = 0.3/1
- 1 - 0.3 = 0.7/2
- 1 - 0.8 = 0.2/3
- 1 - 0.6 = 0.4/4
- 1 - 0.0 = 1.0/5


## 4. SUBTRACTION (A - B)

**Formula:** A ∩ B̄ = min(μA(x), 1 - μB(x))

**Result:** 0.3/1, 0.5/2, 0.2/3, 0.2/4, 1.0/5

**Calculation:**
- min(0.4, 1-0.7=0.3) = 0.3/1

## 5. ALGEBRAIC PRODUCT (A · B)

**Formula:** μA(x) × μB(x)

**Result:** 0.28/1, 0.15/2, 0.56/3, 0.12/4, 0.00/5

**Calculation:**
- 0.4 × 0.7 = 0.28/1

- 0.7 × 0.8 = 0.56/3

## 6. ALGEBRAIC SUM (A ⊕ B)

**Formula:** μA(x) + μB(x) - μA(x) × μB(x)

**Result:** 0.82/1, 0.65/2, 0.94/3, 0.68/4, 1.00/5

- 0.7 + 0.8 - 0.56 = 0.94/3
- 0.2 + 0.6 - 0.12 = 0.68/4

## 7. BOUNDED SUM (A ⊕ B)
