A fuzzy membership function is a mathematical function that defines how each point in the input space is mapped to a membership value between 0 and 1. Unlike classical (crisp) sets where an element either belongs or doesn't belong to a set, fuzzy sets allow partial membership.

<br>

Membership Degree (μ): Ranges from 0 to 1
- μ(x) = 1: Full membership
- μ(x) = 0: No membership
- 0 < μ(x) < 1: Partial membership

<br>

Common Membership Functions:
- Triangular: Simple, linear function
- Trapezoidal: Has a flat plateau
- Gaussian: Bell-shaped curve
- Sigmoidal: S-shaped curve
- Z-shaped: Z-shaped curve (decreasing)
- S-shaped: S-shaped curve (increasing)

```python
import numpy as np
import matplotlib.pyplot as plt

def triangular_mf(x, a, b, c):
    if a <= x <= b:
        return (x - a) / (b - a)
    elif b <= x <= c:
        return (c - x) / (c - b)
    else:
        return 0

def trapezoidal_mf(x, a, b, c, d):
    if a <= x <= b:
        return (x - a) / (b - a)
    elif b <= x <= c:
        return 1
    elif c <= x <= d:
        return (d - x) / (d - c)
    else:
        return 0

def gaussian_mf(x, mean, sigma):
    return np.exp(-((x - mean) ** 2) / (2 * sigma ** 2))

def sigmoidal_mf(x, a, c):
    return 1 / (1 + np.exp(-a * (x - c)))

def z_mf(x, a, b):
    if x <= a:
        return 1
    elif a <= x <= b:
        return 1 - (x - a) / (b - a)
    else:
        return 0

def s_mf(x, a, b):
    if x <= a:
        return 0
    elif a <= x <= b:
        return (x - a) / (b - a)
    else:
        return 1

def plot_membership_functions():
    x = np.linspace(-10, 10, 1000)

    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    fig.suptitle('Different Types of Fuzzy Membership Functions', fontsize=16, fontweight='bold')

    ax1 = axes[0, 0]
    y_triangular = [triangular_mf(xi, -5, 0, 5) for xi in x]
    ax1.plot(x, y_triangular, 'b-', linewidth=2)
    ax1.set_title('Triangular MF\n(a=-5, b=0, c=5)')
    ax1.set_xlabel('x')
    ax1.set_ylabel('Membership μ(x)')
    ax1.grid(True, alpha=0.3)
    ax1.set_ylim(-0.1, 1.1)

    ax2 = axes[0, 1]
    y_trapezoidal = [trapezoidal_mf(xi, -6, -2, 2, 6) for xi in x]
    ax2.plot(x, y_trapezoidal, 'r-', linewidth=2)
    ax2.set_title('Trapezoidal MF\n(a=-6, b=-2, c=2, d=6)')
    ax2.set_xlabel('x')
    ax2.set_ylabel('Membership μ(x)')
    ax2.grid(True, alpha=0.3)
    ax2.set_ylim(-0.1, 1.1)

    ax3 = axes[0, 2]
    y_gaussian = gaussian_mf(x, 0, 2)
    ax3.plot(x, y_gaussian, 'g-', linewidth=2)
    ax3.set_title('Gaussian MF\n(mean=0, sigma=2)')
    ax3.set_xlabel('x')
    ax3.set_ylabel('Membership μ(x)')
    ax3.grid(True, alpha=0.3)
    ax3.set_ylim(-0.1, 1.1)

    ax4 = axes[1, 0]
    y_sigmoidal = sigmoidal_mf(x, 1, 0)
    ax4.plot(x, y_sigmoidal, 'c-', linewidth=2)
    ax4.set_title('Sigmoidal MF\n(a=1, c=0)')
    ax4.set_xlabel('x')
    ax4.set_ylabel('Membership μ(x)')
    ax4.grid(True, alpha=0.3)
    ax4.set_ylim(-0.1, 1.1)

    ax5 = axes[1, 1]
    y_z = [z_mf(xi, -3, 3) for xi in x]
    ax5.plot(x, y_z, 'm-', linewidth=2)
    ax5.set_title('Z-shaped MF\n(a=-3, b=3)')
    ax5.set_xlabel('x')
    ax5.set_ylabel('Membership μ(x)')
    ax5.grid(True, alpha=0.3)
    ax5.set_ylim(-0.1, 1.1)

    ax6 = axes[1, 2]
    y_s = [s_mf(xi, -3, 3) for xi in x]
    ax6.plot(x, y_s, 'orange', linewidth=2)
    ax6.set_title('S-shaped MF\n(a=-3, b=3)')
    ax6.set_xlabel('x')
    ax6.set_ylabel('Membership μ(x)')
    ax6.grid(True, alpha=0.3)
    ax6.set_ylim(-0.1, 1.1)

    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    plot_membership_functions()

```
