## 1. Import Packages

```python
import numpy as np
import pandas as pd
from sklearn import tree, datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

## 2. Load the Dataset

```python
iris = datasets.load_iris()
print(iris.feature_names)
print(iris.target_names)
```

## 3. Separate the independent and dependent variables

```python
x=iris.data
Y = iris.target
```

## 4. Split the data into training and testing sets.

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,Y,test_size=0.3)
```

## 5. Train the model

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(random_state=0, solver='lbfgs', max_iter=1000)
classifier.fit(X_train, Y_train)

predictions = classifier.predict(X_test)
```

## 6. Predict the test data set values using the model above

```python
Y_pred_en=classifier.predict(X_test)
```

## 7. Calculate the accuracy of the model using the accuracy score function

```python
print(("The model accuracy is:"), accuracy_score(Y_test,Y_pred_en)*100)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(Y_test,Y_pred_en)
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test,Y_pred_en))
```

## 8. Testing of the models

```python
Test_values=[[5.4,3.0,4.5,1.5],
             [5.1,3.3,1.7,0.5],
             [6.0,2.2,5.0,1.5]]
print(classifier.predict(Test_values))
```
