```python
import re
import nltk
from pprint import pprint
```

```python
strings = [
    "The Hardest Choices Require The Strongest Wills.",
    "Veni, Vidi, Veci",
    "Welcome to the family Son!",
    "You either die a hero or you live long enough to see yourself become the villain.",
    "Bhaisaab Tear gas ki suvidha nahi hai ka",
    "The right man in the wrong place can make all the difference in the world.",
    "Ah shit, here we go again",
    "Bravo Six Going Dark",
    "Remember No Russian",
    "Heil H"
]

for i, text in enumerate(strings, 1):
    print(f"String {i}: {text}")
```

```python
text1 = strings[0]

print(f"Original: {text1}\n")

# Match uppercase letters
uppercase = re.findall(r'[A-Z]', text1)
print(f"Uppercase letters: {uppercase}")

# Match lowercase letters
lowercase = re.findall(r'[a-z]', text1)
print(f"Lowercase letters: {lowercase}")

# Match all letters
all_letters = re.findall(r'[A-Za-z]', text1)
print(f"All letters: {''.join(all_letters)}")
```

```python
text2 = strings[1]

print(f"Original: {text2}\n")

# * = zero or more
words_star = re.findall(r'[A-Za-z]*', text2)
print(f"* (zero or more): {words_star}")

# + = one or more
words_plus = re.findall(r'[A-Za-z]+', text2)
print(f"+ (one or more): {words_plus}")

# {n} = exact number
words_4 = re.findall(r'\b[A-Za-z]{4}\b', text2)
print(f"{{4}} (exactly 4 letters): {words_4}")

# ? = zero or one
words_comma = re.findall(r'[A-Za-z]+[,]?', text2)
print(f"? (optional comma): {words_comma}")
```

```python
text3 = strings[2]

print(f"Original: {text3}\n")

# ^ = beginning of string
starts_with_welcome = re.match(r'^Welcome', text3)
print(f"Starts with 'Welcome': {bool(starts_with_welcome)}")

# $ = end of string
ends_with_exclamation = re.search(r'!$', text3)
print(f"Ends with '!': {bool(ends_with_exclamation)}")

# Both anchors
full_match = re.match(r'^Welcome.*Son!$', text3)
print(f"Full pattern match: {bool(full_match)}")
```

```python
text4 = strings[3]

print(f"Original: {text4}\n")

# Capture two parts of the quote
pattern = r'(You either die a hero) or (you live long enough to see yourself become the villain)'

matches = re.search(pattern, text4)

if matches:
    print(f"Group 1: {matches.group(1)}")
    print(f"Group 2: {matches.group(2)}")

# Capture 4-letter words
word_capture = re.findall(r'\b([A-Za-z]{4})\b', text4)

print(f"\n4-letter words captured: {word_capture}")
```

```python
text5 = strings[4]

print(f"Original: {text5}\n")

# Match one of several alternatives
pattern = r'(Bhaisaab|Tear gas|suvidha|nahi hai ka)'

matches = re.findall(pattern, text5)

print(f"Matched alternatives: {matches}")

# Check if any alternative exists
has_keyword = re.search(r'gas|ki|ka', text5)

print(f"Contains 'gas' or 'ki' or 'ka': {bool(has_keyword)}")
```

```python
text6 = strings[5]

print(f"Original: {text6}\n")

# Non-capturing group
pattern = r'(?:The|A|An)\s+([A-Za-z]+)'

matches = re.findall(pattern, text6)

print(f"Words after articles: {matches}")

# Capturing group
capturing = re.search(r'(The) (right)', text6)

# Non-capturing group
non_capturing = re.search(r'(?:The) (right)', text6)

print(f"\nCapturing groups: {capturing.groups() if capturing else 'None'}")
print(f"Non-capturing groups: {non_capturing.groups() if non_capturing else 'None'}")
```

```python
text7 = strings[6]

print(f"Original: {text7}\n")

# Positive lookahead
lookahead = re.findall(r'\w+(?= again)', text7)

print(f"Words followed by 'again': {lookahead}")

# Positive lookbehind
lookbehind = re.findall(r'(?<=Ah )\w+', text7)

print(f"Word after 'Ah ': {lookbehind}")

# Combined lookahead + lookbehind
combined = re.findall(r'(?<=here )\w+(?= again)', text7)

print(f"Word between 'here ' and ' again': {combined}")
```

```python
text8 = strings[7]

print(f"Original: {text8}\n")

# Without word boundary
no_boundary = re.findall(r'Bra', text8)

print(f"Without \\b: {no_boundary}")

# With word boundary
with_boundary = re.findall(r'\bBra\b', text8)

print(f"With \\b: {with_boundary}")

# Match all complete words
all_words = re.findall(r'\b\w+\b', text8)

print(f"All whole words: {all_words}")

# Capitalized words
capital_words = re.findall(r'\b[A-Z][a-z]*\b', text8)

print(f"Capitalized words: {capital_words}")
```

```python
text9 = strings[8]

print(f"Original: {text9}\n")

# re.match() - checks from beginning
match_result = re.match(r'Remember', text9)
print(f"re.match('Remember'): {bool(match_result)}")

match_result2 = re.match(r'No', text9)
print(f"re.match('No'): {bool(match_result2)}")

# re.search() - searches anywhere
search_result = re.search(r'Russian', text9)

print(f"re.search('Russian'): {bool(search_result)}")

if search_result:
    print(f"Russian position: {search_result.span()}")

# re.findall() - returns all matches
findall_result = re.findall(r'\b\w+\b', text9)

print(f"re.findall() all words: {findall_result}")
```

```python
text10 = strings[9]

print(f"Original: {text10}\n")

# \s = whitespace
whitespace = re.findall(r'\s', text10)

print(f"Whitespace (\\s): {whitespace}")
print(f"Number of spaces: {len(whitespace)}")

# \S = non-whitespace
non_whitespace = re.findall(r'\S', text10)

print(f"Non-whitespace (\\S): {non_whitespace}")

# \w = word characters
word_chars = re.findall(r'\w', text10)

print(f"Word characters (\\w): {word_chars}")

# \W = non-word characters
non_word = re.findall(r'\W', text10)

print(f"Non-word characters (\\W): {non_word}")

# Extract words
words = re.findall(r'\w+', text10)

print(f"Words extracted: {words}")
```
